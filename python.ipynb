```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,2*np.pi,100)
y=np.sin(x)
y
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,y,label='sin(x)',color='blue', linestyle='-',linewidth=3)
plt.title("Line plot")
plt.xlabel("x values")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
np.random.seed(0)
x=np.random.rand(50)
y=np.random.rand(50)
plt.figure(figsize=(6,4))
plt.scatter(x,y,c='blue',marker='o',alpha=0.5)
plt.title("scatter plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

```python
categories=['A','B','C','D']
values=[10,20,30,40]
plt.figure(figsize=(6,4))
plt.bar(categories,values,color='blue')
plt.title("Bar graph")
plt.xlabel("categorie")
plt.ylabel("values")
plt.grid(False)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data=np.random.randn(1000)
plt.figure(figsize=(6,4))
plt.hist(data,bins=30,color='blue',alpha=0.7,edgecolor='red')
plt.title('Histogram')
plt.xlabel('data values')
plt.ylabel('Frequency')
plt.show()
```

```python
d1=np.random.normal(0,1,100)
d2=np.random.normal(0,1,100)
d3=np.random.normal(-3,1.5,100)
plt.figure(figsize=(6,4))
plt.boxplot([d1,d2,d3],tick_labels=['set1','set2','set3'])
plt.title('boxplot')
plt.ylabel('value')
plt.show()
```

```python
slices=[40,20,10,30]
labels=['A','B','C','D']
explode=(0,0,0.1,0)

import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
plt.pie(slices,labels=labels,explode=explode,autopct="%1.f%%",shadow=True,
        startangle=140)
plt.title("pie chart")
plt.show()
```
